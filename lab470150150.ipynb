## Part 1: Representing a Graph

Before we can search, we need a way to store our problem's structure. For many problems, like navigating a map, finding connections in a social network, or solving a puzzle, the best structure is a **Graph**.

A graph has **Nodes** (also called vertices) and **Edges** that connect them.

To represent a graph in Python, we first need to understand Dictionaries.

### The Adjacency List

An **Adjacency List** is a popular and efficient way to represent a graph. We use a dictionary where:

* **Each key** is a node in the graph.
* **The value** for each key is a **list** of all the nodes it's connected to (its neighbors).

Let's imagine a simple social network graph:



We can represent this graph with the following adjacency list:

```python
social_network_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# To find who 'A' is friends with, we just look up the key 'A'
print(f"A's friends are: {social_network_graph['A']}")

# To find who 'F' is friends with:
print(f"F's friends are: {social_network_graph['F']}")
```

## 2. Breadth-First Search (BFS)

BFS explores a graph **level by level**. It visits the start node, then all of its direct neighbors, then all of their unvisited neighbors, and so on, spreading outward like ripples in water.

BFS uses a **queue** (first in, first out). Python's `collections.deque` is the standard tool for this, `popleft()` removes from the front in O(1) time, unlike a plain list's `pop(0)`.

**Algorithm:**
1. Add the start node to the queue and mark it visited.
2. While the queue is not empty:
   - Remove the node at the front of the queue.
   - Process it (e.g. print it).
   - Add all of its unvisited neighbors to the queue, and mark them visited.
3. Stop when the queue is empty.

```python
from collections import deque

def bfs(graph, start):
    visited = set([start])
    queue = deque([start])
    order = []

    while queue:
        current = queue.popleft()
        order.append(current)
        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order


graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print("BFS order starting from A:", bfs(graph, 'A'))
```

## 5. Depth-First Search (DFS)

DFS can also be done using an explicit **stack** (last in, first out) instead. A plain Python list works fine as a stack, `append()` to push, `pop()` to pop from the end.

| | BFS | DFS |
|---|---|---|
| Data structure | Queue (`deque`, `popleft()`) | Stack (`list`, `pop()`) |
| Removal end | Front | Back |

**Algorithm:**
1. Push the start node onto the stack.
2. While the stack is not empty:
   - Pop a node off the top of the stack.
   - If it hasn't been visited, mark it visited and process it.
   - Push all of its unvisited neighbors onto the stack.
3. Stop when the stack is empty.

### DFS Implementation (Student Exercise)


```python
def dfs_iterative(graph, start):
    visited = set([start])
    stack = [start]
    order = []

    while stack:
        current = stack.pop()
        order.append(current)

        for neighbor in reversed(graph[current]):
            if neighbor not in visited:
                visited.add(neighbor)
                stack.append(neighbor)

    return order

print("DFS (iterative) order starting from A:", dfs_iterative(graph, 'A'))
```
